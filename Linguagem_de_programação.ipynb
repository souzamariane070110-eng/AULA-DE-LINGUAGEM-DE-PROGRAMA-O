```python
filmes = ["1","2","3","4","5"]
print("Bem vindo a classificação do filme")
print("Você tem cinco filmes para classificar")
print("Digite '0' a qualquer momento para parar. \n")

for filme in filmes:
  classificacao = input(f"Como voce classifica '{filmes}' de 1 a 5 (ou 0 para parar): ")

  if classificacao == '0':
    print ("QUE PENA")
    break

  classificacao = int(classificacao)
  if classificacao <1 or classificacao >5:
    print("nota invalida")
  else:
    print(f"voce classificou '{filmes}' com a  '{classificacao}' entrelas.\n")
  print("obrigado")
```
